```python
import cdsapi
from pathlib import Path

target = Path("../data/raw/cams")
target.mkdir(parents=True, exist_ok=True)

dataset = "cams-global-atmospheric-composition-forecasts"
request = {
    "date": ["2020-02-22/2020-02-22"],
    "time": ["00:00"],
    "leadtime_hour": [str(h) for h in range(73)],   # "0", "1", ..., "72"
    "type": ["forecast"],
    "data_format": "netcdf_zip",
    "variable": [
        "particulate_matter_10um",
        "dust_aerosol_optical_depth_550nm",
    ],
    "area": [30, -19, 27, -13],   # North, West, South, East
}

client = cdsapi.Client()
client.retrieve(dataset, request).download(str(target / "cams_2020-02-22_00utc.zip"))
```
